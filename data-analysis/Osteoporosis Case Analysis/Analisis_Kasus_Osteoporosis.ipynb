```python
import plotly.express as px
```

```python
px.histogram(df, 'Age', text_auto=True, title='Distribusi Umur pada Penderita OSteoporosis')
```

```python
px.histogram(df, 'Age', text_auto=True, title='Distribusi Umur pada Penderita OSteoporosis', facet_col='Osteoporosis',  color_discrete_sequence=['#99BC85'])
```

```python
px.histogram(df, 'Age',
             text_auto=True,
             title='Distribusi Umur pada Pasien yang Tidak Menderita dan Menderita Osteoporosis',
             color='Osteoporosis',color_discrete_sequence=['#99BC85', '#BFD8AF'],
             marginal='box' )
```

```python
px.histogram(df, 'Age',
             text_auto=True,
             title='Distribusi Umur dan Asupan Kalsium pada Pasien yang Tidak Menderita dan Menderita Osteoporosis',
             color='Osteoporosis',color_discrete_sequence=['#1B3C73', '#40679E'],
             marginal='box' ,
             facet_col='Calcium Intake')
```
