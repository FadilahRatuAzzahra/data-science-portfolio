# 3. Menganalisis Statistik Data

```python
# 3. Menganalisis Statistik Data
print("\nStatistik Deskriptif:\n", df.describe(include='all'))
```

```python
# Menganalisis Statistik Data Kategorik
df.describe(include=[object])
```

```python
# Menganalisis Statistik Data Numerik
df.describe(include=[float, int])
```

# 4. Mengecek Data Hilang (Missing Values)

```python
# 4. Mengecek Data Hilang (Missing Values)
print("\nJumlah Data Hilang per Kolom:\n", df.isnull().sum())

# Mengisi nilai yang hilang dengan 'Tidak Ada' atau rata-rata (pilih salah satu)
df.fillna('Tidak Ada', inplace=True)
# df.fillna(df.mean(), inplace=True)  # Jika ingin mengisi dengan rata-rata
```

# 5. Mengecek Duplikasi Data

```python
# 5. Mengecek Duplikasi Data
print("\nJumlah Data Duplikat:", df.duplicated().sum())
```

duplikasi data banyak, karena isi kolom nya memang banyak karena merupakan data klasifikasi, dan duplikasi tidak dihapus karena apabila dihapus data akan tidak seimbang

# 6. Mengecek Distribusi Data

```python
# 6. Mengecek Distribusi Data
for col in df.select_dtypes(include=['object']).columns:  # Hanya untuk kolom kategori
    print(f"\nDistribusi Data Kolom '{col}':\n", df[col].value_counts())
```

# 7. Menganalisis Korelasi Antar Kolom Numerik

```python
# 7. Menganalisis Korelasi Antar Kolom Numerik
print("\nMatriks Korelasi:\n", df.corr(numeric_only=True)) # Add numeric_only=True

# Visualisasi Heatmap Korelasi
plt.figure(figsize=(10,6))
sns.heatmap(df.corr(numeric_only=True), annot=True, cmap='coolwarm', fmt=".2f") # Add numeric_only=True
plt.title("Heatmap Korelasi Antar Variabel")
plt.show()
```

# 8. Visualisasi Sederhana

```python
# 8. Visualisasi Sederhana
# Histogram untuk melihat distribusi data numerik
df.hist(figsize=(10, 6), bins=20)
plt.suptitle("Histogram dari Semua Kolom Numerik")
plt.show()

# Boxplot untuk melihat distribusi dan outlier
plt.figure(figsize=(10,6))
sns.boxplot(data=df)
plt.xticks(rotation=90)  # Memutar label agar terbaca
plt.title("Boxplot Distribusi Data")
plt.show()

# Scatter plot jika ada dua kolom numerik
numerik_cols = df.select_dtypes(include=['number']).columns
if len(numerik_cols) >= 2:
    plt.figure(figsize=(8,6))
    sns.scatterplot(x=df[numerik_cols[0]], y=df[numerik_cols[1]])
    plt.title(f"Scatter Plot antara {numerik_cols[0]} dan {numerik_cols[1]}")
    plt.show()
```
