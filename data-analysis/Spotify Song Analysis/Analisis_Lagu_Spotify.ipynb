# Pie Chart

```python
df.groupby(['track_name']).sum()
```

```python
df.groupby(['track_name'])['streams'].sum().to_frame().sort_values(['streams'],  ascending = False).head(5)
```

```python
top_five_song = df.groupby(['track_name'])['streams'].sum().to_frame().sort_values(['streams'],  ascending = False).head(5)
```

```python
top_five_song.index = ['Blinding Lights', 'Shape of You', 'Someone You Loved', 'Dance Monkey', 'Sunflower-Spider-Man']
```

```python
top_five_song.index
```

```python
top_five_song.values.reshape(5)
```

```python
labels = top_five_song.index
values = top_five_song.values.reshape(5)

fig = go.Figure(data=[go.Pie(labels=labels, values=values,
                             textinfo='label+percent',
                             insidetextorientation='radial',
                             marker_colors = ['#99B080' ,'#99BC85', '#BFD8AF', '#D4E7C5', '#E1F0DA']
                            )])
fig.update_layout(
    title_text="Top Five Song by Stream Duration")

fig.show()
```

# Donut Chart

```python
fig = go.Figure(data=[go.Pie(labels=labels,
                             values=values,
                             textinfo='label+percent',
                             insidetextorientation='radial',
                             marker_colors = ['#6E85B7' ,'#82A0D8', '#B2C8DF', '#C4D7E0', '#DAEAF1'],
                             hole = .6
                            )])
fig.update_layout(
    title_text="Top Five Song by Stream Duration",
    # Add annotations in the center of the donut pies.
    annotations=[dict(text='Top 5 Song', x=0.50, y=0.5, font_size=20, showarrow=False)])

fig.show()
```

# Tree Map

```python
top_five_song_id = df.groupby(['track_name'])['streams'].sum().to_frame().sort_values(['streams'],  ascending = False).head(5).index
```

```python
df2 = df.set_index('track_name')
```

```python
top_five_artist = df2.loc[top_five_song_id, 'artist(s)_name'].values
```

```python
df3 = df2.reset_index().set_index('artist(s)_name')
```

```python
df4 = df3.loc[top_five_artist]
```

```python
df4.reset_index(inplace=True)
```

```python
df4
```

```python
fig = px.treemap(df4, path=[px.Constant("Spotify"), 'artist(s)_name', 'track_name'], values='streams',
                  color='streams',
                  color_continuous_scale='teal',
                  color_continuous_midpoint=np.average(df['streams'], weights=df['streams']))
fig.update_layout(margin = dict(t=50, l=25, r=25, b=25),
                  title_text="Top Five Artist by Song and Stream Duration")
fig.show()
```
